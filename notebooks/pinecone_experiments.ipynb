```python
pc.list_indexes()

index = pc.Index("hbface-ilhan")
```

```python
import pickle
import random
import requests



def createData(path):
    with open(path, "rb") as f:
        data = pickle.load(f)

    names = data["names"]
    embeddings = data["embeddings"]

    upserts = []

    response = requests.get('https://api.smart-office.humblebee.ai/api/users')

    if response.status_code == 200:
        users = response.json()

    for name, embedding in zip(names, embeddings):
        user_data = next((user for user in users if user['username'] == name), None)
        upserts.append({
            'id': str(user_data['id']),
            'values': embedding,
            'metadata': {
                'name': user_data['username'],
                'image_path': user_data['image_path'],
                'image_url': user_data['image_url']
            }
        })

    return upserts

upserts = createData('../data/embeddings/ilhan.pkl')
```
