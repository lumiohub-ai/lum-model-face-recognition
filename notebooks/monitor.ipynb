```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV file
df = pd.read_csv("data.csv", parse_dates=["timestamp"])

# Create the plot
plt.figure()
plt.plot(df["timestamp"], df["avg_fps"], label="avg_fps")
plt.plot(df["timestamp"], df["min_fps"], label="min_fps")
plt.plot(df["timestamp"], df["max_fps"], label="max_fps")

# Labels and title
plt.xlabel("Time")
plt.ylabel("FPS")
plt.title("FPS Over Time")
plt.legend()

# Improve layout
plt.xticks(rotation=45)
plt.tight_layout()

# Show the chart
plt.show()
```
